# **Basic Charts And Visual Encoding**


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = {
    'Days': ['Mon','Tues','Wed','Thurs','Fri'],
    'Study_hours': [2,3,2.5,4,3.5],
    'Marks': [50,60,55,70,65],
    'Attendance': [60,70,65,80,75],
    'Sleep_Hours':[6,7,6,8,7],
    'Assignment_Completed': [1,2,1,3,2]
}

df = pd.DataFrame(data)

print(df)
```

## 1. Line Chart (Trend)


```python
# 1. Line Chart (Trend)
plt.plot(df['Days'], df['Study_hours'], marker='o')
plt.title("Soham")
plt.xlabel("Days")
plt.ylabel("Study Hours")
plt.show()
```

### Advance Line Chart


```python
# advance line chart
plt.figure(figsize=(7, 4))

plt.plot(df['Days'], df['Study_hours'],
         marker='o',color='blue',label='Study Hours')

plt.plot(df['Days'], df['Marks'], marker='s',color='red',label='Marks')

plt.title("Soham")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()
plt.show()
```

## 2. Bar Chart (Comparison)

```python
#2. Bar Chart (Comparison)

plt.bar(df['Days'], df['Marks'],color='green')
plt.title("Soham")
plt.xlabel("Days")
plt.ylabel("Marks")
plt.legend()
plt.show()
```

###Advance Bar Graph


```python
#Advance Bar Graph

bars = plt.bar(df['Days'], df['Marks'],color='red')
for bars in bars:
  y = bars.get_height()
  plt.text(bars.get_x() + bars.get_width()/2,y,
           str(y), ha='center',va='bottom')

plt.title("Soham")
plt.xlabel("Days")
plt.ylabel("Marks")
plt.grid(axis='y')
plt.legend()
plt.show()
```

## 3. Histogram (Distribution)


```python
#3. Histogram (Distribution)
#Shows how values are spread

data1 = {
    'Marks': [50,60,55,70,65,75,80,85,90,45,68,72,77,82,88]
    }

df= pd.DataFrame(data1)
plt.hist(df['Marks'],bins=3,edgecolor='black', alpha=1)
plt.title("Soham")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

```

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
import numpy as np

#Example data(more values for better histogram)
Marks= [50,60,55,70,65,75,80,85,90,45,68,72,77,82,88]

#Calculate Mean
mean_value = np.mean(Marks)
plt.figure(figsize=(7,4))
plt.hist(Marks, bins= 6, color= 'skyblue',edgecolor= 'black')

#Mean
plt.axvline(mean_value, color= 'red', linestyle= '--', linewidth= 2, label= 'Mean')
plt.title("Soham")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()

plt.grid(axis= 'y',linestyle='--', alpha=0.5)

plt.show()
```

## 4. Scatter plot


```python
# 4. Scatter plot
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Days': ['Mon','Tues','Wed','Thurs','Fri'],
    'Study_hours': [2,3,2.5,4,3.5],
    'Marks': [50,60,55,70,65],
    'Attendance': [60,70,65,80,75],
    'Sleep_Hours':[6,7,6,8,7],
    'Assignment_Completed': [1,2,1,3,2]
}

df = pd.DataFrame(data)

plt.scatter(df['Study_hours'],df['Marks'])
plt.title("Soham")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
# Example
df['Result'] = ['Fail','Pass','Fail','Pass','Pass']
colors = ['red' if r=='Fail' else 'green'for r in df['Result']]
plt.scatter(df['Study_hours'],df['Marks'],c=colors)
plt.title("Soham")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

# Same Plot Using Seaborn (Better Visualization)

```python
#Dataset-2
data = {
    'Day': ['Mon','Tue','Wed','Thu','Fri','Sat','Sun','Mon','Tue','Wed'],
    'Region': ['North','South','East','West','North','South','East','West','North','South'],
    'Sales': [200,220,250,270,300,350,400,280,260,290],
    'Profit': [20,25,30,35,50,60,80,40,38,45],
    'Customers': [30,35,40,45,50,65,80,55,48,52],
    'Category': ['Food','Clothing','Electronics','Food','Clothing','Electronics','Food',
                 'Clothing','Electronics','Food']
}

df = pd.DataFrame(data)
print(df)
```

### Line Plot


```python
#Line plot
sns.lineplot(x='Day',y='Sales',data=df)
plt.title("Soham")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.show()
```

### Bar Plot

```python
# Bar Plot
sns.barplot(x='Day',y='Profit',data=df)
plt.title("Soham")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

### Histogram


```python
#Histogram
sns.histplot(df['Sales'],bins=5,)
plt.title("Soham")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

### Scatter Plot

```python
#Scatter Plot
sns.scatterplot(x='Sales',y='Profit',data=df)
plt.title("Soham")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```
